```python
import os
import random
import math
from pathlib import Path

import numpy as np
import scipy.io
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, Subset
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# ============================================================
# 1. Reproducibility
# ============================================================
seed = 42
random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)
torch.cuda.manual_seed_all(seed)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ============================================================
# 2. Utility functions
# ============================================================
def set_all_seeds(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

def relative_l2_error_np(gt: np.ndarray, pred: np.ndarray) -> float:
    num = np.linalg.norm(gt - pred)
    den = np.linalg.norm(gt)
    return float(num / (den + 1e-12))

def _resolve_ckpt_path(base_dir: Path, name: str) -> Path:
    ckpt_dir = base_dir / "checkpoints_data_size_study"
    ckpt_dir.mkdir(parents=True, exist_ok=True)
    return ckpt_dir / f"{name}_best.pth"

def _safe_save_same_dir(obj, dest_path: Path):
    tmp_path = dest_path.with_suffix(dest_path.suffix + ".tmp")
    with open(tmp_path, "wb") as f:
        torch.save(obj, f)
    if dest_path.exists():
        try:
            os.remove(dest_path)
        except Exception:
            pass
    os.replace(str(tmp_path), str(dest_path))
    return str(dest_path)

# ============================================================
# 3. Load data
# ============================================================
data_path = "Burger.mat"
data_dir = Path(os.path.abspath(data_path)).parent

mat = scipy.io.loadmat(data_path)
u_all = mat["input"]          # [N, 100]
s_all = mat["output"]         # [N, T, X]
s_all = s_all[:, :100, :]     # keep first 100 time steps

N, T, X = s_all.shape
assert u_all.shape[0] == N and u_all.shape[1] == X == 100, "Unexpected data shape"

print("u_all shape:", u_all.shape)
print("s_all shape:", s_all.shape)

# ============================================================
# 4. Fixed train / val / test split
# ============================================================
X_trainval, X_test, y_trainval, y_test = train_test_split(
    u_all, s_all, test_size=0.20, random_state=seed
)
X_train, X_val, y_train, y_val = train_test_split(
    X_trainval, y_trainval, test_size=0.125, random_state=seed
)
# Final split:
# train = 70%
# val   = 10%
# test  = 20%

print("Train:", X_train.shape, y_train.shape)
print("Val  :", X_val.shape, y_val.shape)
print("Test :", X_test.shape, y_test.shape)

# ============================================================
# 5. Normalize using FULL training set statistics
# ============================================================
mu_ic  = X_train.mean(axis=0, keepdims=True)
std_ic = X_train.std(axis=0, keepdims=True) + 1e-8

mu_sol  = y_train.mean(axis=(0, 1), keepdims=True)
std_sol = y_train.std(axis=(0, 1), keepdims=True) + 1e-8

X_train_n = (X_train - mu_ic) / std_ic
X_val_n   = (X_val   - mu_ic) / std_ic
X_test_n  = (X_test  - mu_ic) / std_ic

y_train_n = (y_train - mu_sol) / std_sol
y_val_n   = (y_val   - mu_sol) / std_sol
y_test_n  = (y_test  - mu_sol) / std_sol

# ============================================================
# 6. Coordinates
# ============================================================
grid = np.linspace(0.0, 1.0, 100)
Xg, Tg = np.meshgrid(grid, grid, indexing='xy')
coords_np = np.stack([Xg, Tg], axis=-1).astype(np.float32)   # [100,100,2]

# ============================================================
# 7. Dataset / dataloader
# ============================================================
class Burgers1DDataset(Dataset):
    def __init__(self, ic, sol, coords):
        self.ic = torch.tensor(ic, dtype=torch.float32)
        self.sol = torch.tensor(sol, dtype=torch.float32)
        self.coords = torch.tensor(coords, dtype=torch.float32)

    def __len__(self):
        return self.ic.shape[0]

    def __getitem__(self, i):
        return self.ic[i], self.sol[i]

train_ds_full = Burgers1DDataset(X_train_n, y_train_n, coords_np)
val_ds        = Burgers1DDataset(X_val_n,   y_val_n,   coords_np)
test_ds       = Burgers1DDataset(X_test_n,  y_test_n,  coords_np)

coords = train_ds_full.coords.to(device)

# ============================================================
# 8. Models
# ============================================================
class FourierEmbedding(nn.Module):
    def __init__(self, input_dim, mapping_size=64, sigma=10.0):
        super().__init__()
        B = torch.randn(input_dim, mapping_size) * sigma
        self.register_buffer("B", B)

    def forward(self, x):
        x_proj = 2 * math.pi * x @ self.B
        return torch.cat([torch.sin(x_proj), torch.cos(x_proj)], dim=-1)

class TrunkFourier(nn.Module):
    def __init__(self, input_dim=2, mapping_size=64, sigma=10.0, hidden=256, out_features=100):
        super().__init__()
        self.embed = FourierEmbedding(input_dim=input_dim, mapping_size=mapping_size, sigma=sigma)
        self.net = nn.Sequential(
            nn.Linear(2 * mapping_size, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, out_features)
        )

    def forward(self, coords):
        x = coords.view(-1, 2)
        x = self.embed(x)
        return self.net(x)

class TrunkMLP(nn.Module):
    def __init__(self, hidden=256, out_features=100):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, out_features)
        )

    def forward(self, coords):
        x = coords.view(-1, 2)
        return self.net(x)

class Branch(nn.Module):
    def __init__(self, out_features=100):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(100, 50), nn.ReLU(),
            nn.Linear(50, 50), nn.ReLU(),
            nn.Linear(50, 50), nn.ReLU(),
            nn.Linear(50, 50), nn.ReLU(),
            nn.Linear(50, 50), nn.ReLU(),
            nn.Linear(50, out_features)
        )

    def forward(self, ic):
        return self.net(ic)

class DeepONet(nn.Module):
    def __init__(self, trunk, branch_out_features=100):
        super().__init__()
        self.branch = Branch(out_features=branch_out_features)
        self.trunk = trunk

    def forward(self, ic, coords):
        B = ic.size(0)
        G = coords.size(0)  # 100

        trunk_out  = self.trunk(coords)   # [10000,100]
        branch_out = self.branch(ic)      # [B,100]

        y = torch.matmul(branch_out, trunk_out.T)  # [B,10000]
        return y.view(B, G, G)                     # [B,100,100]

# ============================================================
# 9. Denormalization / evaluation
# ============================================================
def denorm_sol(y_norm: torch.Tensor) -> torch.Tensor:
    mu = torch.tensor(mu_sol, dtype=torch.float32, device=y_norm.device)
    sd = torch.tensor(std_sol, dtype=torch.float32, device=y_norm.device)
    return y_norm * sd + mu

@torch.no_grad()
def evaluate_mean_rel_l2(model, loader, device, coords):
    model.eval()
    rels = []

    for ic, gt in loader:
        ic = ic.to(device)
        gt = gt.to(device)

        pred = model(ic, coords)

        pred_d = denorm_sol(pred).cpu().numpy()
        gt_d   = denorm_sol(gt).cpu().numpy()

        for i in range(gt_d.shape[0]):
            rels.append(relative_l2_error_np(gt_d[i], pred_d[i]))

    return float(np.mean(rels))

@torch.no_grad()
def val_rel_l2_global(model, loader, device, coords):
    model.eval()
    num_sq, den_sq = 0.0, 0.0

    for ic, gt in loader:
        ic = ic.to(device)
        gt = gt.to(device)

        pred = model(ic, coords)

        pred_d = denorm_sol(pred)
        gt_d   = denorm_sol(gt)

        e = pred_d - gt_d
        num_sq += torch.sum(e ** 2).item()
        den_sq += torch.sum(gt_d ** 2).item()

    return math.sqrt(num_sq / (den_sq + 1e-20))

# ============================================================
# 10. Train functions
# ============================================================
def train_one_epoch(model, loader, optimizer, criterion, device, coords):
    model.train()
    running_loss = 0.0

    for ic, gt in loader:
        ic = ic.to(device)
        gt = gt.to(device)

        optimizer.zero_grad()
        pred = model(ic, coords)
        loss = criterion(pred, gt)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * ic.size(0)

    return running_loss / len(loader.dataset)

def fit_model(model, name, train_loader, val_loader, coords, base_dir,
              max_epochs=1000, lr=1e-3, patience=150):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode="min", factor=0.5, patience=30
    )

    best_val = float("inf")
    best_path = _resolve_ckpt_path(base_dir, name)
    wait = 0

    hist = {"train_loss": [], "val_rel_l2": []}

    print(f"[INFO] Saving checkpoint to: {best_path}")

    for epoch in range(1, max_epochs + 1):
        tr_loss = train_one_epoch(model, train_loader, optimizer, criterion, device, coords)
        val_r2 = val_rel_l2_global(model, val_loader, device, coords)

        hist["train_loss"].append(tr_loss)
        hist["val_rel_l2"].append(val_r2)
        scheduler.step(val_r2)

        print(f"[{name}] Epoch {epoch:4d}/{max_epochs} | Train MSE(n): {tr_loss:.6e} | Val Rel-L2: {val_r2:.6e}")

        if val_r2 < best_val - 1e-6:
            best_val = val_r2
            _safe_save_same_dir(
                {
                    "epoch": epoch,
                    "model_state": model.state_dict(),
                    "best_val_rel_l2": best_val
                },
                best_path
            )
            wait = 0
        else:
            wait += 1
            if wait >= patience:
                print(f"[{name}] Early stopping at epoch {epoch}. Best Val Rel-L2: {best_val:.6e}")
                break

    return str(best_path), hist

# ============================================================
# 11. Model builders
# ============================================================
def build_deeponet():
    return DeepONet(
        trunk=TrunkMLP(hidden=256, out_features=100),
        branch_out_features=100
    ).to(device)

def build_fedonet(mapping_size=64, sigma=10.0):
    return DeepONet(
        trunk=TrunkFourier(
            input_dim=2,
            mapping_size=mapping_size,
            sigma=sigma,
            hidden=256,
            out_features=100
        ),
        branch_out_features=100
    ).to(device)

# ============================================================
# 12. Data-size sweep
# ============================================================
def make_subset_loader(dataset, n_samples, batch_size=128, shuffle=True):
    indices = np.arange(len(dataset))[:n_samples]
    subset = Subset(dataset, indices.tolist())
    loader = DataLoader(subset, batch_size=batch_size, shuffle=shuffle, drop_last=False)
    return loader, subset

def run_data_size_study(
    train_dataset,
    val_dataset,
    test_dataset,
    coords,
    train_sizes,
    fed_mapping_size=64,
    fed_sigma=10.0,
    batch_size=128,
    max_epochs=1000,
    lr=1e-3,
    patience=150,
):
    val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, drop_last=False)
    test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, drop_last=False)

    results = {
        "train_size": [],
        "deeponet_val_rel_l2": [],
        "deeponet_test_rel_l2": [],
        "fedonet_val_rel_l2": [],
        "fedonet_test_rel_l2": [],
    }

    for n_train in train_sizes:
        print("\n" + "=" * 100)
        print(f"Training with n_train = {n_train}")
        print("=" * 100)

        train_loader, _ = make_subset_loader(
            train_dataset, n_samples=n_train, batch_size=batch_size, shuffle=True
        )

        # ---------------- DeepONet ----------------
        set_all_seeds(seed)
        deeponet = build_deeponet()

        deep_ckpt, _ = fit_model(
            model=deeponet,
            name=f"deeponet_n{n_train}",
            train_loader=train_loader,
            val_loader=val_loader,
            coords=coords,
            base_dir=data_dir,
            max_epochs=max_epochs,
            lr=lr,
            patience=patience
        )

        deep_ckpt_data = torch.load(deep_ckpt, map_location=device)
        deeponet.load_state_dict(deep_ckpt_data["model_state"])

        deep_best_val = float(deep_ckpt_data["best_val_rel_l2"])
        deep_test_rel = evaluate_mean_rel_l2(deeponet, test_loader, device, coords)

        print(f"[DeepONet | n={n_train}] Best Val Rel-L2 : {deep_best_val:.6e}")
        print(f"[DeepONet | n={n_train}] Test Mean Rel-L2: {deep_test_rel:.6e}")

        # ---------------- FEDONet ----------------
        set_all_seeds(seed)
        fedonet = build_fedonet(mapping_size=fed_mapping_size, sigma=fed_sigma)

        fed_ckpt, _ = fit_model(
            model=fedonet,
            name=f"fedonet_n{n_train}",
            train_loader=train_loader,
            val_loader=val_loader,
            coords=coords,
            base_dir=data_dir,
            max_epochs=max_epochs,
            lr=lr,
            patience=patience
        )

        fed_ckpt_data = torch.load(fed_ckpt, map_location=device)
        fedonet.load_state_dict(fed_ckpt_data["model_state"])

        fed_best_val = float(fed_ckpt_data["best_val_rel_l2"])
        fed_test_rel = evaluate_mean_rel_l2(fedonet, test_loader, device, coords)

        print(f"[FEDONet  | n={n_train}] Best Val Rel-L2 : {fed_best_val:.6e}")
        print(f"[FEDONet  | n={n_train}] Test Mean Rel-L2: {fed_test_rel:.6e}")

        results["train_size"].append(n_train)
        results["deeponet_val_rel_l2"].append(deep_best_val)
        results["deeponet_test_rel_l2"].append(deep_test_rel)
        results["fedonet_val_rel_l2"].append(fed_best_val)
        results["fedonet_test_rel_l2"].append(fed_test_rel)

        del deeponet, fedonet
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    return results

# ============================================================
# 13. Choose training sizes
# ============================================================
n_train_full = len(train_ds_full)
print("Full training set size:", n_train_full)

# Use absolute training sizes
train_sizes = [50, 100, 300, 400, 500, 800, 900]
train_sizes = [n for n in train_sizes if n <= n_train_full]

# If your dataset is smaller than some requested sizes, they are removed.
print("Training sizes used:", train_sizes)

# ============================================================
# 14. Run the study
# ============================================================
results = run_data_size_study(
    train_dataset=train_ds_full,
    val_dataset=val_ds,
    test_dataset=test_ds,
    coords=coords,
    train_sizes=train_sizes,
    fed_mapping_size=512,   # adjust if you want
    fed_sigma=10.0,
    batch_size=128,
    max_epochs=1000,       # reduce if runtime is too high
    lr=1e-3,
    patience=15000
)

# ============================================================
# 15. Summary table in terminal
# ============================================================
train_sizes_arr = np.array(results["train_size"])
deep_val_arr    = np.array(results["deeponet_val_rel_l2"])
deep_test_arr   = np.array(results["deeponet_test_rel_l2"])
fed_val_arr     = np.array(results["fedonet_val_rel_l2"])
fed_test_arr    = np.array(results["fedonet_test_rel_l2"])

print("\n" + "=" * 110)
print("FINAL SUMMARY: Training Data Size vs Rel-L2 Error")
print("=" * 110)
for n, dv, dt, fv, ft in zip(train_sizes_arr, deep_val_arr, deep_test_arr, fed_val_arr, fed_test_arr):
    print(
        f"n_train={n:4d} | "
        f"DeepONet  Val={dv:.6e}, Test={dt:.6e} | "
        f"FEDONet  Val={fv:.6e}, Test={ft:.6e}"
    )

# ============================================================
# 16. Plot test Rel-L2 vs training size
# ============================================================
plt.figure(figsize=(8, 5))
plt.plot(train_sizes_arr, deep_test_arr, marker='o', linewidth=2, label='DeepONet Test Rel-L2')
plt.plot(train_sizes_arr, fed_test_arr, marker='s', linewidth=2, label='FEDONet Test Rel-L2')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Number of Training Samples', fontsize=13)
plt.ylabel('Mean Test Relative L2 Error', fontsize=13)
plt.title('Effect of Training Data Size on Test Rel-L2 Error', fontsize=14)
plt.grid(True, which='both', linestyle='--', alpha=0.4)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

# ============================================================
# 17. Plot validation Rel-L2 vs training size
# ============================================================
plt.figure(figsize=(8, 5))
plt.plot(train_sizes_arr, deep_val_arr, marker='o', linewidth=2, label='DeepONet Best Val Rel-L2')
plt.plot(train_sizes_arr, fed_val_arr, marker='s', linewidth=2, label='FEDONet Best Val Rel-L2')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Number of Training Samples', fontsize=13)
plt.ylabel('Best Validation Relative L2 Error', fontsize=13)
plt.title('Effect of Training Data Size on Validation Rel-L2 Error', fontsize=14)
plt.grid(True, which='both', linestyle='--', alpha=0.4)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

# ============================================================
# 18. Optional combined plot with percentage of full training set
# ============================================================
train_percent = 100.0 * train_sizes_arr / n_train_full

plt.figure(figsize=(8, 5))
plt.plot(train_percent, deep_test_arr, marker='o', linewidth=2, label='DeepONet Test Rel-L2')
plt.plot(train_percent, fed_test_arr, marker='s', linewidth=2, label='FEDONet Test Rel-L2')
plt.yscale('log')
plt.xlabel('Training Set Size (% of Full Training Set)', fontsize=13)
plt.ylabel('Mean Test Relative L2 Error', fontsize=13)
plt.title('Test Rel-L2 vs Fraction of Training Data', fontsize=14)
plt.grid(True, which='both', linestyle='--', alpha=0.4)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```
