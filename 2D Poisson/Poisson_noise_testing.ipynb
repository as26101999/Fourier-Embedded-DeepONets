```python
# =============================== Poisson DeepONet / FEDONet ===============================
# Train on clean data, then test robustness to noisy input RHS f(x,y)
# Compares:
#   (1) Vanilla DeepONet
#   (2) Fourier-Embedded DeepONet (FEDONet)
# Plots mean relative L2 error vs input noise level

import os
import random
import math
import numpy as np
import scipy.io
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split

# ------------------------------- Reproducibility -------------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

# --------------------------------- Utilities ----------------------------------
def rel_l2(a: torch.Tensor, b: torch.Tensor, eps: float = 1e-12):
    """
    Relative L2 per-sample: ||a-b||_2 / (||b||_2 + eps)
    Accepts tensors shaped [B, 1, H, W] or [B, H, W].
    Returns tensor of shape [B].
    """
    if a.dim() == 4:
        a = a.view(a.size(0), -1)
        b = b.view(b.size(0), -1)
    elif a.dim() == 3:
        a = a.view(a.size(0), -1)
        b = b.view(b.size(0), -1)
    num = torch.norm(a - b, dim=1)
    den = torch.norm(b, dim=1) + eps
    return num / den

@torch.no_grad()
def evaluate_loader_rel_l2(model, loader, device, coords):
    model.eval()
    total_rel, count = 0.0, 0
    coords = coords.to(device)
    for f_batch, u_batch in loader:
        f_batch = f_batch.to(device)
        u_batch = u_batch.to(device)
        preds = model(f_batch, coords)
        r = rel_l2(preds, u_batch)  # [B]
        total_rel += r.sum().item()
        count += r.numel()
    return total_rel / max(count, 1)

@torch.no_grad()
def evaluate_loader_mse(model, loader, device, coords):
    model.eval()
    crit = nn.MSELoss(reduction='sum')
    total_loss, count = 0.0, 0
    coords = coords.to(device)
    for f_batch, u_batch in loader:
        f_batch = f_batch.to(device)
        u_batch = u_batch.to(device)
        preds = model(f_batch, coords)
        total_loss += crit(preds, u_batch).item()
        count += u_batch.numel()
    return total_loss / max(count, 1)

def add_gaussian_noise_relative(x, noise_level, rng=None):
    """
    Add Gaussian noise to each sample x_i with std = noise_level * rms(x_i).

    x           : numpy array [N, H, W]
    noise_level : scalar, e.g. 0.05 means 5% relative noise
    """
    if rng is None:
        rng = np.random.default_rng(SEED)

    x_noisy = x.copy()
    flat = x.reshape(x.shape[0], -1)
    rms = np.sqrt(np.mean(flat**2, axis=1, keepdims=True))  # [N,1]
    noise = rng.standard_normal(size=flat.shape).astype(np.float32)
    flat_noisy = flat + noise_level * rms * noise
    x_noisy = flat_noisy.reshape(x.shape).astype(np.float32)
    return x_noisy

# --------------------------------- Dataset ------------------------------------
class PoissonDataset(Dataset):
    """
    Stores pairs (rhs_f, sol_u) with shapes [N, H, W] each.
    Provides tensors [N, 1, H, W] and a static coord grid [H, W, 2] in [0,1]^2.
    """
    def __init__(self, rhs_f, sol_u):
        assert rhs_f.shape == sol_u.shape, "rhs_f and sol_u must have same shape"
        self.rhs_f = torch.tensor(rhs_f, dtype=torch.float32).unsqueeze(1)  # [N,1,H,W]
        self.sol_u = torch.tensor(sol_u, dtype=torch.float32).unsqueeze(1)  # [N,1,H,W]

        H, W = self.rhs_f.shape[-2:]
        grid_x = np.linspace(0.0, 1.0, W, dtype=np.float32)
        grid_y = np.linspace(0.0, 1.0, H, dtype=np.float32)
        X, Y = np.meshgrid(grid_x, grid_y)
        coords = np.stack([X, Y], axis=-1)  # [H,W,2]
        self.coords = torch.tensor(coords, dtype=torch.float32)

    def __len__(self):
        return self.rhs_f.shape[0]

    def __getitem__(self, idx):
        return self.rhs_f[idx], self.sol_u[idx]

# --------------------------------- Models -------------------------------------
class BranchNet(nn.Module):
    """
    CNN encoder for the input field f(x,y).
    """
    def __init__(self, out_dim=128):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1, 16, kernel_size=5, stride=2, padding=2), nn.ReLU(),
            nn.Conv2d(16, 32, kernel_size=5, stride=2, padding=2), nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=5, stride=2, padding=2), nn.ReLU(),
            nn.AdaptiveAvgPool2d((16, 16)),
            nn.Flatten(),
            nn.Linear(64 * 16 * 16, 128), nn.ReLU(),
            nn.Linear(128, out_dim),
        )

    def forward(self, x):
        return self.net(x)

class TrunkNetVanilla(nn.Module):
    """
    Vanilla trunk MLP: (x,y) -> R^out_dim
    """
    def __init__(self, out_dim=128):
        super().__init__()
        self.mlp = nn.Sequential(
            nn.Linear(2, 128), nn.ReLU(),
            # nn.Linear(256, 256), nn.ReLU(),
            # nn.Linear(128, 128), nn.ReLU(),
            # nn.Linear(128, 128), nn.ReLU(),
            nn.Linear(128, 128), nn.ReLU(),
            nn.Linear(128, out_dim), nn.ReLU(),
        )

    def forward(self, coords_flat):
        return self.mlp(coords_flat)

class FourierEmbedding(nn.Module):
    """
    Random Fourier Features for coordinates (x,y).
    """
    def __init__(self, input_dim=2, mapping_size=256, sigma=10.0):
        super().__init__()
        B = torch.randn(input_dim, mapping_size) * sigma
        self.register_buffer("B", B)

    def forward(self, x):
        x_proj = 2.0 * math.pi * (x @ self.B)
        return torch.cat([torch.sin(x_proj), torch.cos(x_proj)], dim=-1)

class TrunkNetFourier(nn.Module):
    """
    Fourier-feature trunk MLP: (x,y) -> R^out_dim
    """
    def __init__(self, out_dim=128, mapping_size=256, sigma=10.0):
        super().__init__()
        self.embed = FourierEmbedding(input_dim=2, mapping_size=mapping_size, sigma=sigma)
        in_dim = 2 * mapping_size
        self.mlp = nn.Sequential(
            nn.Linear(in_dim, 128), nn.ReLU(),
            nn.Linear(128, 128), nn.ReLU(),
            # nn.Linear(256, 256), nn.ReLU(),
            # nn.Linear(256, 256), nn.ReLU(),
            # nn.Linear(256, 256), nn.ReLU(),
            nn.Linear(128, out_dim), nn.ReLU()
        )

    def forward(self, coords_flat):
        z = self.embed(coords_flat)
        return self.mlp(z)

class DeepONet(nn.Module):
    """
    Generic DeepONet = branch(f) dot trunk(coords)
    """
    def __init__(self, branch: nn.Module, trunk: nn.Module):
        super().__init__()
        self.branch = branch
        self.trunk = trunk

    def forward(self, f_batch, coords):
        """
        f_batch: [B,1,H,W]
        coords : [H,W,2]
        returns: [B,1,H,W]
        """
        B, _, H, W = f_batch.shape
        coords_flat = coords.view(-1, 2)              # [H*W, 2]
        trunk_out   = self.trunk(coords_flat)         # [H*W, D]
        branch_out  = self.branch(f_batch)            # [B, D]
        out = torch.matmul(branch_out, trunk_out.t()) # [B, H*W]
        out = out.view(B, 1, H, W)
        return out

# ----------------------------- Training / Checkpoint ---------------------------
def train_model(
    model, train_loader, val_loader, device, coords,
    num_epochs=500, lr=1e-3, ckpt_path="model_best.pt", print_every=50
):
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.MSELoss()

    best_val_rel_l2 = float("inf")
    history = {"train_mse": [], "val_rel_l2": []}
    coords = coords.to(device)

    for epoch in range(1, num_epochs + 1):
        model.train()
        train_loss_sum, train_count = 0.0, 0

        for f_batch, u_batch in train_loader:
            f_batch = f_batch.to(device)
            u_batch = u_batch.to(device)

            optimizer.zero_grad()
            preds = model(f_batch, coords)
            loss = criterion(preds, u_batch)
            loss.backward()
            optimizer.step()

            train_loss_sum += loss.item() * u_batch.numel()
            train_count += u_batch.numel()

        train_mse = train_loss_sum / max(train_count, 1)
        history["train_mse"].append(train_mse)

        val_rel = evaluate_loader_rel_l2(model, val_loader, device, coords)
        history["val_rel_l2"].append(val_rel)

        if val_rel < best_val_rel_l2:
            best_val_rel_l2 = val_rel
            torch.save(
                {
                    "model_state": model.state_dict(),
                    "best_val_rel_l2": best_val_rel_l2,
                    "epoch": epoch
                },
                ckpt_path
            )

        if (epoch % print_every) == 0 or epoch == 1 or epoch == num_epochs:
            print(f"[{epoch:5d}/{num_epochs}] "
                  f"train MSE: {train_mse:.6e} | val relL2: {val_rel:.6e} "
                  f"| best: {best_val_rel_l2:.6e}")

    return history, best_val_rel_l2

# ------------------------- Noise Robustness Evaluation -------------------------
@torch.no_grad()
def evaluate_noise_robustness(
    model,
    X_test_clean_norm,
    y_test_clean_norm,
    coords,
    device,
    batch_size=256,
    noise_levels=None,
    seed=123
):
    """
    Evaluates a trained model on noisy versions of the normalized test input.
    The target remains the clean normalized solution.
    """
    if noise_levels is None:
        noise_levels = [0.0, 0.01, 0.02, 0.05, 0.10, 0.15, 0.20]

    rng = np.random.default_rng(seed)
    results = []

    for nl in noise_levels:
        X_test_noisy = add_gaussian_noise_relative(X_test_clean_norm, nl, rng=rng)
        noisy_ds = PoissonDataset(rhs_f=X_test_noisy, sol_u=y_test_clean_norm)
        noisy_loader = DataLoader(noisy_ds, batch_size=batch_size, shuffle=False, drop_last=False)

        mean_rel = evaluate_loader_rel_l2(model, noisy_loader, device, coords)
        results.append(mean_rel)
        print(f"Noise level = {100*nl:5.1f}% | Mean rel L2 = {mean_rel:.6e}")

    return np.array(results, dtype=np.float64)

# ---------------------------------- Main --------------------------------------
if __name__ == "__main__":
    # ------------------------- Load data (.mat) -------------------------
    data = scipy.io.loadmat("poisson_1.mat")

    # Expected keys: 'f_data' (RHS), 'phi_data' (solution)
    rhs_all = data["f_data"].astype(np.float32)    # [N,H,W]
    sol_all = data["phi_data"].astype(np.float32)  # [N,H,W]

    print("rhs_all shape:", rhs_all.shape)
    print("sol_all shape:", sol_all.shape)

    # ------------------ Train/Val/Test split (no leakage) ------------------
    X_train_full, X_test, y_train_full, y_test = train_test_split(
        rhs_all, sol_all, test_size=0.10, random_state=SEED, shuffle=True
    )

    X_train, X_val, y_train, y_val = train_test_split(
        X_train_full, y_train_full, test_size=0.10, random_state=SEED, shuffle=True
    )

    # ------------------------ Normalize using TRAIN only ------------------------
    f_mean, f_std = X_train.mean(), X_train.std()
    u_mean, u_std = y_train.mean(), y_train.std()

    def norm_f(x): return (x - f_mean) / (f_std + 1e-12)
    def norm_u(x): return (x - u_mean) / (u_std + 1e-12)
    def denorm_u(x): return x * (u_std + 1e-12) + u_mean

    X_train = norm_f(X_train)
    X_val   = norm_f(X_val)
    X_test  = norm_f(X_test)

    y_train = norm_u(y_train)
    y_val   = norm_u(y_val)
    y_test  = norm_u(y_test)

    # ------------------------- Datasets & Loaders -------------------------
    train_ds = PoissonDataset(rhs_f=X_train, sol_u=y_train)
    val_ds   = PoissonDataset(rhs_f=X_val,   sol_u=y_val)
    test_ds  = PoissonDataset(rhs_f=X_test,  sol_u=y_test)

    BATCH_SIZE = 256
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=False)
    val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, drop_last=False)
    test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, drop_last=False)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    coords = train_ds.coords.to(device)

    print("Using device:", device)

    # --------------------------- Vanilla DeepONet ---------------------------
    branch_v = BranchNet(out_dim=128)
    trunk_v  = TrunkNetVanilla(out_dim=128)
    model_v  = DeepONet(branch_v, trunk_v).to(device)

    print("\n=== Training VANILLA DeepONet ===")
    hist_v, best_val_rel_v = train_model(
        model_v,
        train_loader,
        val_loader,
        device,
        coords,
        num_epochs=500,
        lr=1e-3,
        ckpt_path="vanilla_best.pt",
        print_every=50
    )

    ckpt_v = torch.load("vanilla_best.pt", map_location=device)
    model_v.load_state_dict(ckpt_v["model_state"])
    test_rel_v = evaluate_loader_rel_l2(model_v, test_loader, device, coords)
    print(f"[VANILLA] Best val rel L2: {best_val_rel_v:.6e} | Clean test mean rel L2: {test_rel_v:.6e}")

    # ---------------------- Fourier-Embedded DeepONet ----------------------
    branch_f = BranchNet(out_dim=128)
    trunk_f  = TrunkNetFourier(out_dim=128, mapping_size=256, sigma=10.0)
    model_f  = DeepONet(branch_f, trunk_f).to(device)

    print("\n=== Training FOURIER-EMBEDDED DeepONet (FEDONet) ===")
    hist_f, best_val_rel_f = train_model(
        model_f,
        train_loader,
        val_loader,
        device,
        coords,
        num_epochs=500,
        lr=1e-3,
        ckpt_path="fourier_best.pt",
        print_every=50
    )

    ckpt_f = torch.load("fourier_best.pt", map_location=device)
    model_f.load_state_dict(ckpt_f["model_state"])
    test_rel_f = evaluate_loader_rel_l2(model_f, test_loader, device, coords)
    print(f"[FEDONET] Best val rel L2: {best_val_rel_f:.6e} | Clean test mean rel L2: {test_rel_f:.6e}")

    # ---------------------- Noise robustness experiment ----------------------
    noise_levels = [0.00, 0.01, 0.02, 0.05, 0.10, 0.15, 0.20, 0.30]

    print("\n=== Noise robustness: VANILLA DeepONet ===")
    rel_errors_v = evaluate_noise_robustness(
        model=model_v,
        X_test_clean_norm=X_test,
        y_test_clean_norm=y_test,
        coords=coords,
        device=device,
        batch_size=BATCH_SIZE,
        noise_levels=noise_levels,
        seed=123
    )

    print("\n=== Noise robustness: FEDONet ===")
    rel_errors_f = evaluate_noise_robustness(
        model=model_f,
        X_test_clean_norm=X_test,
        y_test_clean_norm=y_test,
        coords=coords,
        device=device,
        batch_size=BATCH_SIZE,
        noise_levels=noise_levels,
        seed=123
    )

    # ----------------------------- Print summary -----------------------------
    print("\n==================== Summary ====================")
    print(f"Clean Test relL2 - Vanilla DeepONet : {test_rel_v:.6e}")
    print(f"Clean Test relL2 - FEDONet          : {test_rel_f:.6e}")
    print("\nNoise robustness table:")
    for nl, ev, ef in zip(noise_levels, rel_errors_v, rel_errors_f):
        print(f"Noise = {100*nl:5.1f}% | Vanilla = {ev:.6e} | FEDONet = {ef:.6e}")

    # ----------------------------- Plot results -----------------------------
    noise_percent = 100 * np.array(noise_levels)

    plt.figure(figsize=(8, 6))
    plt.plot(noise_percent, rel_errors_v, marker='o', linewidth=2, label='DeepONet')
    plt.plot(noise_percent, rel_errors_f, marker='s', linewidth=2, label='FEDONet')
    plt.xlabel('Noise level added to input f (%)', fontsize=13)
    plt.ylabel('Mean relative L2 error', fontsize=13)
    plt.title('Effect of input noise on Poisson operator learning models', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=12)
    plt.tight_layout()
    plt.show()

    # ---------------------- Optional: save the figure -----------------------
    plt.figure(figsize=(8, 6))
    plt.plot(noise_percent, rel_errors_v, marker='o', linewidth=2, label='DeepONet')
    plt.plot(noise_percent, rel_errors_f, marker='s', linewidth=2, label='FEDONet')
    plt.xlabel('Noise level added to input f (%)', fontsize=13)
    plt.ylabel('Mean relative L2 error', fontsize=13)
    plt.title('Rel L2 error vs input noise', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=12)
    plt.tight_layout()
    plt.savefig("relL2_vs_noise_poisson_deeponet_fedonet.png", dpi=300)
    plt.close()

    print("\nSaved figure: relL2_vs_noise_poisson_deeponet_fedonet.png")
    print("Saved checkpoints: vanilla_best.pt, fourier_best.pt")
```

```python
# ============================================================
# NOISE TEST VISUALIZATION (Poisson DeepONet vs FEDONet)
# Same arrangement pattern as requested
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import torch

noise_level = 0.10   # 10% noise
seed_vis = 123

# ------------------------------------------------------------
# Add noise to CLEAN normalized test input X_test
# ------------------------------------------------------------
rng = np.random.default_rng(seed_vis)

# X_test here is already normalized in your training script
# Noise is added relative to per-sample RMS
flat = X_test.reshape(X_test.shape[0], -1)
rms = np.sqrt(np.mean(flat**2, axis=1, keepdims=True)) + 1e-12
noise = rng.standard_normal(size=flat.shape).astype(np.float32) * (noise_level * rms)

X_test_noisy = (flat + noise).reshape(X_test.shape).astype(np.float32)

# ------------------------------------------------------------
# Predict using noisy input
# ------------------------------------------------------------
model_v.eval()
model_f.eval()

coords_t = coords.to(device)

X_noisy_t = torch.tensor(X_test_noisy, dtype=torch.float32).unsqueeze(1).to(device)  # [N,1,H,W]

with torch.no_grad():
    PRED_van_noisy = model_v(X_noisy_t, coords_t).squeeze(1).cpu().numpy()   # [N,H,W]
    PRED_fed_noisy = model_f(X_noisy_t, coords_t).squeeze(1).cpu().numpy()   # [N,H,W]

GT_test_noisy = y_test.copy()   # clean target, still normalized

# ------------------------------------------------------------
# Compute Rel L2 errors
# ------------------------------------------------------------
rel2_fed_noisy = np.array([
    np.linalg.norm(GT_test_noisy[i] - PRED_fed_noisy[i]) /
    (np.linalg.norm(GT_test_noisy[i]) + 1e-12)
    for i in range(GT_test_noisy.shape[0])
])

rel2_van_noisy = np.array([
    np.linalg.norm(GT_test_noisy[i] - PRED_van_noisy[i]) /
    (np.linalg.norm(GT_test_noisy[i]) + 1e-12)
    for i in range(GT_test_noisy.shape[0])
])

print(f"Mean Rel-L2 ({100*noise_level:.0f}% noise)")
print("FEDONet :", rel2_fed_noisy.mean())
print("DeepONet:", rel2_van_noisy.mean())

# ------------------------------------------------------------
# Choose sample
# You can change this selection rule if you want
# ------------------------------------------------------------
idx = int(np.argmin(rel2_fed_noisy))

input_clean = X_test
input_noisy = X_test_noisy

full_solution = GT_test_noisy
outputs_fed = PRED_fed_noisy
outputs_van = PRED_van_noisy

# ------------------------------------------------------------
# Grid
# ------------------------------------------------------------
H, W = full_solution.shape[-2], full_solution.shape[-1]
x = np.linspace(0, 1, W)
y = np.linspace(0, 1, H)
XX, YY = np.meshgrid(x, y, indexing='xy')

# ------------------------------------------------------------
# Errors
# ------------------------------------------------------------
error_fed = full_solution[idx] - outputs_fed[idx]
error_van = full_solution[idx] - outputs_van[idx]

vmin_sol = min(full_solution[idx].min(), outputs_fed[idx].min(), outputs_van[idx].min())
vmax_sol = max(full_solution[idx].max(), outputs_fed[idx].max(), outputs_van[idx].max())

vmin_err = min(error_fed.min(), error_van.min())
vmax_err = max(error_fed.max(), error_van.max())

vmin_inp = min(input_clean[idx].min(), input_noisy[idx].min())
vmax_inp = max(input_clean[idx].max(), input_noisy[idx].max())

# ============================================================
# Plot
# ============================================================

plt.rcParams.update({
    'font.size': 22,
    'axes.titlesize': 28,
    'axes.labelsize': 24,
    'xtick.labelsize': 20,
    'ytick.labelsize': 20,
    'legend.fontsize': 22
})

plt.rcParams['image.cmap'] = 'jet'
fig = plt.figure(figsize=(25, 14))

# ---- Noisy input RHS ----
ax0 = fig.add_subplot(241)
c0 = ax0.pcolor(XX, YY, input_noisy[idx], vmin=vmin_inp, vmax=vmax_inp)
ax0.set_title(f'Noisy Input')
fig.colorbar(c0, ax=ax0)

# ---- Exact solution ----
ax1 = fig.add_subplot(242)
c1 = ax1.pcolor(XX, YY, full_solution[idx], vmin=vmin_sol, vmax=vmax_sol)
ax1.set_title('Exact Solution')
fig.colorbar(c1, ax=ax1)

# ---- DeepONet prediction ----
ax2 = fig.add_subplot(243)
c2 = ax2.pcolor(XX, YY, outputs_van[idx], vmin=vmin_sol, vmax=vmax_sol)
ax2.set_title('DeepONet')
fig.colorbar(c2, ax=ax2)

# ---- DeepONet error ----
ax3 = fig.add_subplot(244)
c3 = ax3.pcolor(XX, YY, error_van, vmin=vmin_err, vmax=vmax_err)
ax3.set_title('DeepONet Error')
fig.colorbar(c3, ax=ax3)

# ---- Clean input RHS ----
ax4 = fig.add_subplot(245)
c4 = ax4.pcolor(XX, YY, input_clean[idx], vmin=vmin_inp, vmax=vmax_inp)
ax4.set_title('Clean Input')
fig.colorbar(c4, ax=ax4)

# ---- Exact solution again (same arrangement style as your pattern) ----
ax5 = fig.add_subplot(246)
c5 = ax5.pcolor(XX, YY, full_solution[idx], vmin=vmin_sol, vmax=vmax_sol)
ax5.set_title('Exact Solution')
fig.colorbar(c5, ax=ax5)

# ---- FEDONet prediction ----
ax6 = fig.add_subplot(247)
c6 = ax6.pcolor(XX, YY, outputs_fed[idx], vmin=vmin_sol, vmax=vmax_sol)
ax6.set_title('FEDONet')
fig.colorbar(c6, ax=ax6)

# ---- FEDONet error ----
ax7 = fig.add_subplot(248)
c7 = ax7.pcolor(XX, YY, error_fed, vmin=vmin_err, vmax=vmax_err)
ax7.set_title('FEDONet Error')
fig.colorbar(c7, ax=ax7)

plt.tight_layout()
plt.show()
```
